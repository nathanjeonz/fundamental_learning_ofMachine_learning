```python
import pandas as pd
penguins = pd.read_csv('data/penguins_size.csv')
penguins.head()
```

```python
px.scatter(x=penguins['body_mass_g'], y=penguins['flipper_length_mm'], title='Penguin Body Mass vs Flipper Length', labels={'x': 'Body Mass (g)', 'y': 'Flipper Length (mm)'})
```

```python
# country total specis population
country_species_population = penguins.groupby('island')['species'].count().reset_index()
country_species_population.columns = ['Island', 'Population']
```

```python
px.bar(x=country_species_population['Island'], y=country_species_population['Population'], title='Penguin Population by Island', labels={'x': 'Island', 'y': 'Population'})
```
