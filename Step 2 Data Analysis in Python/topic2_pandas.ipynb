```python
fruits = ("apple","orange","banana","kiwi","dragonfruit","cherry","kiwi","lychee")
prices = [100,200,150,250,350,215,250,300]

data = [fruits,prices]

data
```

```python
df['new_pprice'] = df['prices'] * 1.10
```
